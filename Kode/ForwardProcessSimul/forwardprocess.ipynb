```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

img = Image.open("Magnusfiksv2.png").convert("RGB")
img = np.array(img) / 255.0

steps = 10
beta_start = 0.0001
beta_end = 0.70
betas = np.linspace(beta_start, beta_end, steps)

x = img.copy()
images = [x]

for beta in betas:
    noise = np.random.normal(0, 1, x.shape)
    x = np.sqrt(1 - beta) * x + np.sqrt(beta) * noise
    images.append(np.clip(x, 0, 1))

# velg hvilke steg du vil vise
show_steps = [0, 3, 5, 10]

fig, axes = plt.subplots(1, len(show_steps), figsize=(10,3))

for i, step in enumerate(show_steps):
    axes[i].imshow(images[step])
    axes[i].set_title(f"Step {step}")
    axes[i].axis("off")

plt.savefig("forward_diffusion_magnus.png", dpi=300, bbox_inches="tight")
plt.show()
```
